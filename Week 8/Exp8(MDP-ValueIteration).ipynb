```python
import numpy as np

size = 3
goal = (2, 2)
trap = (1, 1)
gamma = 0.9

actions = [(-1,0),(1,0),(0,-1),(0,1)]

def next_state(s, a):
    i, j = s[0] + a[0], s[1] + a[1]
    i = max(0, min(i, size-1))
    j = max(0, min(j, size-1))
    return (i, j)

def reward(s):
    if s == goal: return 0
    if s == trap: return -10
    return -1

V = {(i,j): 0 for i in range(size) for j in range(size)}

while True:
    delta = 0
    for s in V:
        if s == goal or s == trap:
            continue

        v = V[s]
        V[s] = max(reward(next_state(s,a)) + gamma * V[next_state(s,a)] for a in actions)
        delta = max(delta, abs(v - V[s]))

    if delta < 0.01:
        break

print("Value Iteration:")
for s in V:
    print(s, round(V[s], 2))
```

```python
import numpy as np
import random

size = 3
goal = (2, 2)
trap = (1, 1)
gamma = 0.9

# Actions with names
actions = {
    "UP": (-1, 0),
    "DOWN": (1, 0),
    "LEFT": (0, -1),
    "RIGHT": (0, 1)
}

def next_state(s, a):
    dr, dc = actions[a]
    i, j = s[0] + dr, s[1] + dc
    i = max(0, min(i, size - 1))
    j = max(0, min(j, size - 1))
    return (i, j)

def reward(s):
    if s == goal:
        return 0
    if s == trap:
        return -10
    return -1

states = [(i, j) for i in range(size) for j in range(size)]
V = {s: 0 for s in states}

# Random initial policy (using action names)
policy = {s: random.choice(list(actions.keys()))
          for s in states if s not in [goal, trap]}

while True:
    # --- Policy Evaluation ---
    while True:
        delta = 0
        for s in states:
            if s in [goal, trap]:
                continue

            v = V[s]
            a = policy[s]
            ns = next_state(s, a)

            V[s] = reward(ns) + gamma * V[ns]
            delta = max(delta, abs(v - V[s]))

        if delta < 0.01:
            break

    # --- Policy Improvement ---
    stable = True
    for s in policy:
        old = policy[s]

        policy[s] = max(
            actions.keys(),
            key=lambda a: reward(next_state(s, a)) + gamma * V[next_state(s, a)]
        )

        if old != policy[s]:
            stable = False

    if stable:
        break

# --- Output ---
print("Policy Iteration Result:\n")
for s in policy:
    ns = next_state(s, policy[s])
    print(f"{s} -> {policy[s]} -> {ns}, Value: {round(V[s], 2)}")
```
