```python
import heapq

# Goal state
GOAL = ((1,2,3),
        (4,5,6),
        (7,8,0))

# Manhattan Distance Heuristic
def heuristic(state):
    distance = 0
    for i in range(3):
        for j in range(3):
            value = state[i][j]
            if value != 0:
                goal_x = (value - 1) // 3
                goal_y = (value - 1) % 3
                distance += abs(i - goal_x) + abs(j - goal_y)
    return distance


# Find blank position
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j


# Generate neighbors
def get_neighbors(state):
    neighbors = []
    x, y = find_blank(state)

    moves = {
        "UP": (-1, 0),
        "DOWN": (1, 0),
        "LEFT": (0, -1),
        "RIGHT": (0, 1)
    }

    for move, (dx, dy) in moves.items():
        nx, ny = x + dx, y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [list(row) for row in state]
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            neighbors.append((tuple(map(tuple, new_state)), move))

    return neighbors


# A* Algorithm
def solve_puzzle(start):
    heap = []
    visited = set()

    g = 0
    h = heuristic(start)
    heapq.heappush(heap, (g + h, g, start, []))

    while heap:
        f, g, current, path = heapq.heappop(heap)

        if current == GOAL:
            return path, g

        if current in visited:
            continue

        visited.add(current)

        for neighbor, move in get_neighbors(current):
            if neighbor not in visited:
                heapq.heappush(heap, (
                    g + 1 + heuristic(neighbor),
                    g + 1,
                    neighbor,
                    path + [move]
                ))

    return None


# Initial State (same as example)
START = ((1,2,3),
         (4,0,6),
         (7,5,8))

path, moves = solve_puzzle(START)

print("Path of blank moves:")
print(" -> ".join(path))
print("Total moves:", moves)
```
